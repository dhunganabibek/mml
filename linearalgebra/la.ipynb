```python
# importing all the required libraries
import torch
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
# calculating the norm
a  = np.array([1, 2, 3])

norm_manual = (a[0]**2 + a[1]**2 + a[2]**2)**0.5
print("Manual norm:", norm_manual)

norm_numpy = np.linalg.norm(a)
print("NumPy norm:", norm_numpy)
```

```python
# check the orthogonal 
a = np.array([1, 0])
b = np.array([0, 1])
dot_product = np.dot(a, b)
print("Dot product:", dot_product)  
a@b
```

```python
# matrix tensors
X = np.array([[1, 2, 3], [4, 5, 6]])
print("NumPy array:\n", X)

print("Shape:", X.shape)
print("Data type:", X.dtype)
print("Number of dimensions:", X.ndim)
print("Size:", X.size)
```

```python
# working with pytorch
X_pt = torch.tensor([[1, 2, 3], [4, 5, 6]])
print("PyTorch tensor:\n", X_pt)

print("Shape:", X_pt.shape)
print("Data type:", X_pt.dtype)
print("Number of dimensions:", X_pt.ndim)
print("Size:", X_pt.numel())
```

```python
# working with tensor flow
X_tf = tf.Variable([[1, 2, 3], [4, 5, 6]])
print("TensorFlow variable:\n", X_tf)

print("Shape:", X_tf.shape)
print("Data type:", X_tf.dtype)

```

```python
# calculation of hadamard product
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])
c = a * b
print("Hadamard product:\n", c)
```

```python
X = torch.tensor([[1, 2, 3], [4, 5, 6]])
X.sum(0) # axis 0: sum down the rows
X.sum(1) # axis 1: sum across the columns
```

```python
# matrix multiplication
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
C = np.dot(A, B)
print("Matrix multiplication result from numpy:\n", C)


A_pt = torch.from_numpy(A)
B_pt = torch.from_numpy(B)
C_pt = torch.matmul(A_pt, B_pt)
print("Matrix multiplication result from PyTorch:\n", C_pt)

A_pt @ B_pt
```

```python
# Frobenius mean (square of sum of values and square root)
A = np.array([[1, 2], [3, 4]])

frobenius_norm = np.linalg.norm(A, 'fro')
print("Frobenius norm:", frobenius_norm)

frobenius_norm_pt = torch.linalg.norm(torch.from_numpy(A).float(), 'fro')
print("Frobenius norm (PyTorch):", frobenius_norm_pt)

# calculate manually
T = torch.from_numpy(A).float()
T.abs().square().sum().sqrt().item()

```
