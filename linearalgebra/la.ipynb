```python
import numpy as np
import matplotlib.pyplot as plt
from typing import Optional, Union, Sequence

def plotVector(
    vectors: Union[Sequence[float], Sequence[Sequence[float]]],
    *,
    color: str = 'tab:blue',
    label: Optional[str] = None,
    pivot: str = 'tail',          # 'tail' | 'mid' | 'tip'
    scale: float = 1,             # 1 => U,V in data units
    width: Optional[float] = 0.005,
    headwidth: Optional[float] = 6,
    headlength: Optional[float] = 8,
    equal: bool = True,           # keep x/y aspect equal
    annotate: bool = False,       # place text labels near arrow tips
    ax: Optional[plt.Axes] = None,
    **kwargs                      # any extra plt.quiver kwargs
):
    """
    Plot one or many vectors using Matplotlib quiver.

    Parameters
    ----------
    vectors : array-like
        Either a single [x, y, u, v] or an array of shape (N, 4).
        - x,y are bases
        - u,v are components
    color : str
        Arrow color for the whole set.
    label : str | None
        Legend label for this set. Use ax.legend() to display.
    pivot : 'tail' | 'mid' | 'tip'
        Where the arrow is anchored relative to (x, y).
    scale : float
        Quiver scale. With angles='xy', scale_units='xy': vector ends at (x+u, y+v) when scale=1.
    width, headwidth, headlength : float
        Arrow style.
    equal : bool
        Set equal aspect ratio.
    annotate : bool
        If True and label provided:
          - for a single vector: put the label near the tip
          - for multiple vectors: if label is a list of length N, annotate each.
    ax : matplotlib.axes.Axes | None
        Target axes. Defaults to current axes.
    kwargs : dict
        Forwarded to ax.quiver (e.g., alpha=..., zorder=..., linewidth=...).

    Returns
    -------
    Q : matplotlib.quiver.Quiver
        The quiver artist.
    """
    ax = ax if ax is not None else plt.gca()

    arr = np.asarray(vectors, dtype=float)
    if arr.ndim == 1:
        if arr.size != 4:
            raise ValueError("Single vector must be a 1D array/list of length 4: [x, y, u, v].")
        arr = arr.reshape(1, 4)
    elif arr.ndim == 2:
        if arr.shape[1] != 4:
            raise ValueError("vectors must have shape (N, 4): columns are [x, y, u, v].")
    else:
        raise ValueError("Unsupported vectors shape. Provide [x,y,u,v] or (N,4).")

    X, Y, U, V = arr[:, 0], arr[:, 1], arr[:, 2], arr[:, 3]

    Q = ax.quiver(
        X, Y, U, V,
        angles='xy', scale_units='xy', scale=scale,
        color=color, pivot=pivot,
        width=width, headwidth=headwidth, headlength=headlength,
        label=label,
        **kwargs
    )

    if equal:
        ax.set_aspect('equal', 'box')

    # Optional text annotations
    if annotate and label is not None:
        # If label is a single string and there are many vectors, annotate only the first
        if isinstance(label, str):
            # place near the first vector tip
            xt, yt = X[0] + U[0], Y[0] + V[0]
            ax.text(xt, yt, f" {label}", va='center', ha='left', color=color)
        else:
            # assume an iterable of per-vector labels
            labels = np.asarray(label, dtype=object)
            if labels.shape[0] != X.shape[0]:
                raise ValueError("When annotate=True and label is a list, it must have length N.")
            for xi, yi, ui, vi, li in zip(X, Y, U, V, labels):
                xt, yt = xi + ui, yi + vi
                ax.text(xt, yt, f" {li}", va='center', ha='left', color=color)

    return Q


if __name__ == "__main__":
    # Example 1: single vector
    fig, ax = plt.subplots(figsize=(5, 5))
    plotVector([0, 0, 2, 1], color='tab:blue', label='(2,1)', ax=ax, annotate=True)
    ax.set_xlim(-0.5, 2.5); ax.set_ylim(-0.5, 2.0)
    ax.grid(True); ax.legend(); plt.show()

    # Example 2: multiple vectors (your arrays)
    import numpy as np
    X = np.array([0, 1, 2])
    Y = np.array([0, 1, 0])
    U = np.array([1, 0, -1])
    V = np.array([0, 1,  1])
    vectors = np.column_stack([X, Y, U, V])

    fig, ax = plt.subplots(figsize=(5, 5))
    plotVector(vectors, color='tab:orange', label='set A', pivot='mid', ax=ax)
    ax.set_xlim(-1, 3); ax.set_ylim(-1, 3)
    ax.grid(True); ax.legend(); plt.show()
```

```python
# define the plotting function
from typing import Sequence

def plotVectors(
    *vectors: Sequence[float],
    xlim: Sequence[float] = (-10, 10),
    ylim: Sequence[float] = (-10, 10)
):
    arr = np.array([list(v) for v in vectors], dtype=float)
    X, Y, U, V = arr[:, 0], arr[:, 1], arr[:, 2], arr[:, 3]
    ax =  plt.gca()
    Q = ax.quiver(
        X, Y, U, V,
        angles='xy', scale_units='xy', scale=1
    )

    ax.set_aspect('equal', 'box')
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
    ax.grid(True)
    
```

```python
from PIL import Image
import numpy as np

# Load image (grayscale)
img = Image.open('image.png').convert('L')
A = np.array(img)

# SVD
U, s, Vh = np.linalg.svd(A, full_matrices=False)

# Compression: keep only k singular values
def compress_img_np(U, s, Vh, k):
    U_k = U[:, :k]
    s_k = s[:k]
    Vh_k = Vh[:k, :]
    S_k = np.diag(s_k)
    A_k = U_k @ S_k @ Vh_k
    return np.clip(A_k, 0, 255).astype(np.uint8)

k = 50  # Compression level
A_compressed = compress_img_np(U, s, Vh, k)

# Save compressed image
Image.fromarray(A_compressed).save('compressed_numpy.png')
```

```python
#doing image compression with svd: torch
img = Image.open('image.png').convert('L')
A = np.array(img)
A = torch.tensor(A, dtype=torch.float32)

# SVD
U, s, Vh = torch.linalg.svd(A)

# Compression: keep only k singular values
def compress_img_torch(U, s, Vh, k):
    U_k = U[:, :k]
    s_k = s[:k]
    Vh_k = Vh[:k, :]
    A_k = U_k @ torch.diag(s_k) @ Vh_k
    return torch.clamp(A_k, 0, 255).byte()

k = 50  # Compression level
A_compressed = compress_img_torch(U, s, Vh, k)

# Save compressed image
Image.fromarray(A_compressed.numpy()).save('compressed_torch.png')
```

```python
# Moore Penrose pseudo inverse
def moore_penrose_pseudo_inverse(A):
    U, s, Vh = np.linalg.svd(A)
    S_inv = np.zeros_like(A.T)
    for i in range(len(s)):
        if s[i] > 1e-10:  # Threshold to avoid division by zero
            S_inv[i, i] = 1 / s[i]
    return Vh.T @ S_inv @ U.T
```
